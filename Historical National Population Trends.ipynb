# Historical National Population Trends

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("POPH.csv", low_memory=False)
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)
```

```python
df.head()
```

```python
df.info()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(
    df["date"],
    df["value"],
    label="National Population",
    color="purple",
    linewidth=2.5,
)
plt.title(
    "Historical National Population Growth Over Time",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Population (Persons)", fontsize=12)
plt.legend()
plt.show()
```

```python
df["Population_Growth_Rate"] = df["value"].pct_change() * 100
```

```python
plt.figure(figsize=(12, 5))
plt.plot(
    df["date"],
    df["Population_Growth_Rate"],
    label="Annual Growth Rate (%)",
    color="teal",
)
plt.title(
    "Annual National Population Growth Rate", fontsize=14, fontweight="bold"
)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Growth Rate (%)", fontsize=12)
plt.legend()
plt.show()
```
